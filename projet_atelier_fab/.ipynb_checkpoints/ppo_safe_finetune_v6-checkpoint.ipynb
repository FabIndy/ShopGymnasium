```python
# Cellule 1 — Imports de base et configuration

import numpy as np
import torch
import gymnasium as gym
from gymnasium import spaces

from sb3_contrib import MaskablePPO
from sb3_contrib.common.wrappers import ActionMasker

from stable_baselines3.common.callbacks import EvalCallback

from env.workshop_env import WorkshopEnv  # ton environnement atelier

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
```

```python
# Cellule 2 — Environnement d'entraînement avec masque d'actions

def mask_fn(env: WorkshopEnv):
    """
    Fonction utilisée par ActionMasker pour récupérer le masque des actions.
    """
    return env.get_action_mask()

# Environnement enveloppé pour PPO
env_train = ActionMasker(WorkshopEnv(), mask_fn)

print("Espace d'observation :", env_train.observation_space)
print("Espace d'actions     :", env_train.action_space)
```

```python
# ======================================================
# Cellule 3 — Chargement modèle DAgger + BASELINE AVANT PPO
# ======================================================

from sb3_contrib import MaskablePPO

MODEL_PATH = "student_dagger_expertv4_final.zip"

# Chargement du modèle issu de DAgger
model_ppo = MaskablePPO.load(
    MODEL_PATH,
    env=env_train,
    device="cuda"
)

print("Modèle DAgger chargé depuis", MODEL_PATH)
```

```python
# ======================================================
# Cellule 4 — Évaluation robuste sur N semaines (SEEDS FIXES)
# ======================================================

def evaluate_model_weekly(model, seeds, max_steps=10080):
    """
    Évalue le modèle sur des semaines complètes avec seeds contrôlées.
    """
    rewards = []

    for i, seed in enumerate(seeds):
        env = WorkshopEnv()
        obs, _ = env.reset(seed=seed)

        total_reward = 0.0

        for t in range(max_steps):
            mask = env.get_action_mask()
            action, _ = model.predict(
                obs,
                deterministic=True,
                action_masks=mask
            )
            obs, r, terminated, truncated, _ = env.step(action)
            total_reward += r

            if terminated or truncated:
                break

        rewards.append(total_reward)
        print(f"Seed {seed} — reward = {total_reward:.2f}")

    rewards = np.array(rewards, dtype=np.float32)
    print(f"\nReward moyen : {rewards.mean():.2f} ± {rewards.std():.2f}")
    return rewards
```

```python
# ======================================================
# Cellule 5 — BASELINE AVANT PPO (DAgger pur)
# ======================================================

BASE_SEED = 123
N_EVAL_WEEKS = 20    # <-- CHANGÉ (était 5)

EVAL_SEEDS = [BASE_SEED + i for i in range(N_EVAL_WEEKS)]

print("=== Évaluation AVANT PPO (policy issue de DAgger) ===")
rewards_before = evaluate_model_weekly(
    model_ppo,
    seeds=EVAL_SEEDS
)
```

```python
# ======================================================
# Cellule 6 — Configuration PPO ROBUSTE (APRÈS baseline)
# ======================================================

from stable_baselines3.common.utils import get_schedule_fn

# Learning rate légèrement augmenté mais sûr
NEW_LR = 1e-4

for param_group in model_ppo.policy.optimizer.param_groups:
    param_group["lr"] = NEW_LR

model_ppo.learning_rate = NEW_LR

# Clip range conservateur
model_ppo.clip_range = get_schedule_fn(0.1)

# Exploration fortement contrôlée
model_ppo.ent_coef = 0.003

# Exploitation un peu plus poussée par batch
model_ppo.n_epochs = min(model_ppo.n_epochs + 2, 15)

print("\n=== Hyperparamètres PPO configurés ===")
print("learning_rate =", NEW_LR)
print("clip_range    = 0.1 (constant)")
print("ent_coef      =", model_ppo.ent_coef)
print("n_epochs      =", model_ppo.n_epochs)
print("n_steps       =", model_ppo.n_steps)
print("batch_size    =", model_ppo.batch_size)
print("================================================\n")
```

```python
# ======================================================
# Cellule 6-B — Environnement d’évaluation + EvalCallback
# ======================================================

# Environnement d'évaluation (séparé de l'entraînement)
eval_env = ActionMasker(WorkshopEnv(), mask_fn)

# Callback d'évaluation PPO (plus robuste)
eval_callback = EvalCallback(
    eval_env,
    n_eval_episodes=10,             # <-- CHANGÉ (était 5)
    eval_freq=50_000,
    best_model_save_path="./ppo_safe_best_v6",
    deterministic=True,
    render=False
)

print("EvalCallback initialisé (10 épisodes / évaluation)")
```

```python
# Cellule 7 — Entraînement PPO avec contraintes "safe"
from datetime import datetime

TOTAL_TIMESTEPS = 1_000_000
LOG_NAME = f"ppo_safe_finetune_v6_{datetime.now().strftime('%Y%m%d_%H%M')}"

print(f"=== Entraînement PPO (safe) pour {TOTAL_TIMESTEPS} timesteps ===")

model_ppo.set_env(env_train) # par sécurité

model_ppo.learn( total_timesteps=TOTAL_TIMESTEPS, tb_log_name=LOG_NAME, callback=eval_callback, progress_bar=True )

print("=== Fin de l'entraînement PPO (safe) ===")
```

```python
# ======================================================
# Cellule 7-B — Chargement du best model PPO (ou fallback)
# ======================================================

BEST_MODEL_PATH = "./ppo_safe_best_v6/best_model.zip"

try:
    best_model = MaskablePPO.load(
        BEST_MODEL_PATH,
        env=env_train,        # <-- AJOUT IMPORTANT
        device="cuda"
    )
    print(f"Best model chargé depuis : {BEST_MODEL_PATH}")
except Exception as e:
    print(" Impossible de charger le best model, fallback sur model_ppo.")
    print("Erreur :", e)
    best_model = model_ppo
```

```python
# ======================================================
# Cellule 8 — Évaluation APRÈS PPO (SEEDS IDENTIQUES)
# ======================================================

print("=== Évaluation APRÈS PPO ===")
rewards_after = evaluate_model_weekly(
    best_model,
    seeds=EVAL_SEEDS
)
```

```python
# ======================================================
# cellule 9 : Sauvegarde propre du modèle PPO finetuné
# ======================================================

import os
from datetime import datetime

# Dossier de sauvegarde
SAVE_DIR = "ppo_final_model"
os.makedirs(SAVE_DIR, exist_ok=True)

# Nom du fichier = date + heure pour versioning automatique
timestamp = datetime.now().strftime("%Y%m%d_%H%M")
MODEL_PATH = os.path.join(SAVE_DIR, f"ppo_finetuned_{timestamp}.zip")

# 1) Sauvegarde du modèle complet
best_model.save(MODEL_PATH)

# 2) Sauvegarde des hyperparamètres dans un fichier texte
with open(os.path.join(SAVE_DIR, "hyperparams_v6.txt"), "w") as f:
    f.write("=== PPO Finetuned Hyperparameters ===\n")
    f.write(f"learning_rate = {best_model.learning_rate}\n")
    f.write(f"clip_range    = constant schedule 0.1\n")
    f.write(f"ent_coef      = {best_model.ent_coef}\n")
    f.write(f"n_steps       = {best_model.n_steps}\n")
    f.write(f"batch_size    = {best_model.batch_size}\n")
    f.write(f"timestamp     = {timestamp}\n")

# 3) Message confirmation
print("======================================")
print("Modèle PPO finetuné sauvegardé avec succès !")
print("Chemin :", MODEL_PATH)
print("Hyperparamètres enregistrés dans : ppo_final_model/hyperparams_v6.txt")
print("======================================")
```

```python
# ======================================================
# Cellule 10 — Sauvegarde des rewards d'évaluation PPO
# ======================================================

import pandas as pd
import os

# Récupération des évaluations stockées par EvalCallback
eval_log_path = "./ppo_safe_best_v6/evaluations.npz"

if not os.path.exists(eval_log_path):
    raise FileNotFoundError("Fichier evaluations.npz introuvable.")

data = np.load(eval_log_path)

df_eval = pd.DataFrame({
    "timesteps": data["timesteps"],
    "mean_reward": data["results"].mean(axis=1),
    "std_reward": data["results"].std(axis=1)
})

df_eval.to_csv("ppo_eval_curve_v6.csv", index=False)

print("Courbe d'évaluation sauvegardée → ppo_eval_curve_v6.csv")
```
